# Langkah 2: Mengkodekan fitur ordinal

Pada langkah ini, Anda akan menggunakan fungsi mapper untuk mengubah fitur ordinal menjadi nilai numerik terurut.

Mulailah dengan mendapatkan jenis kolom baru dari DataFrame:


```python
df_car.info()
```

Pertama, tentukan nilai apa yang terdapat dalam kolom ordinal. 

Dimulai dengan fitur **num-of-doors**, Anda dapat menggunakan `value_counts` untuk menemukan nilai-nilai.

```python
df_car['num-of-doors'].value_counts()
```

Fitur ini hanya memiliki dua nilai: *empat* dan *dua*. Anda dapat membuat mapper sederhana yang berisi kamus:

```python
door_mapper = {"two": 2,
              "four": 4}
```

Anda kemudian dapat menggunakan metode `replace` dari pandas untuk menghasilkan kolom numerik baru berdasarkan kolom **num-of-doors**.

```python
df_car['doors'] = df_car["num-of-doors"].replace(door_mapper)
```

Ketika Anda menampilkan DataFrame, Anda akan melihat kolom baru di sebelah kanan. Ini berisi representasi numerik dari jumlah pintu.


```python
df_car.head()
```

Ulangi proses dengan kolom **num-of-cylinders**.

Pertama, dapatkan nilainya.

```python
df_car['num-of-cylinders'].value_counts()
```

Selanjutnya, buat mapper.

```python
cylinder_mapper = {"two":2,
                  "three":3,
                  "four":4,
                  "five":5,
                  "six":6,
                  "eight":8,
                  "twelve":12}
```

Menerapkan mapper dengan menggunakan metode `replace`.


```python
df_car['cylinders'] = df_car['num-of-cylinders'].replace(cylinder_mapper)
```

```python
df_car.head()
```

Untuk informasi lebih lanjut tentang metode `replace`, lihat [pandas.DataFrame.replace](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.replace.html) dalam dokumentasi pandas.

# Langkah 3: Mengkodekan data kategoris non-ordinal

Pada langkah ini, Anda akan mengkodekan data non-ordinal dengan menggunakan metode `get_dummies` dari pandas.

Dua fitur yang tersisa tidak ordinal. 

Menurut deskripsi atribut, nilai berikut mungkin dilakukan:

- aspiration: std, turbo. 
- roda penggerak: 4wd, fwd, rwd. 

Anda mungkin berpikir bahwa strategi yang tepat adalah mengkonversi nilai-nilai ini menjadi nilai numerik. Misalnya, pertimbangkan fitur **drive-wheels**. Anda bisa menggunakan *4wd = 1*, *fwd = 2*, dan *rwd = 3*. Namun, *fwd* tidak kurang dari *rwd*. Nilai-nilai ini tidak memiliki urutan, tetapi Anda baru saja memasukannya dengan menetapkan nilai numerik ini.

Strategi yang benar adalah mengubah nilai-nilai ini menjadi *fitur biner* untuk setiap nilai dalam fitur asli. Proses ini sering disebut *pengkode one-hot* dalam machine learning, atau *dummying* dalam statistik.

pandas menyediakan metode `get_dummies`, yang mengubah data menjadi fitur biner. Untuk informasi lebih lanjut, lihat [pandas.get_dummies](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.get_dummies.html) dalam dokumentasi pandas.



Menurut deskripsi atribut, **drive-wheels** memiliki tiga nilai yang mungkin.

```python
df_car['drive-wheels'].value_counts()
```

Gunakan metode `get_dummies` untuk menambahkan fitur biner baru ke DataFrame.


```python
df_car = pd.get_dummies(df_car,columns=['drive-wheels'])
```

```python
df_car.head()
```

Ketika Anda memeriksa set data, Anda akan melihat tiga kolom baru di sebelah kanan: 

- **drive-wheels_4wd**
- **drive-wheels_fwd**
- **drive-wheels_rwd**

Pengkodeannya bersifat langsung Jika nilai di kolom **drive-wheels** adalah *4wd*, maka *1* adalah nilai di kolom **drive-wheels_4wd**. *0* adalah nilai untuk kolom lain yang dihasilkan. Jika nilai dalam kolom **drive-wheels** adalah *fwd*, maka *1* adalah nilai di kolom **drive-wheels_fwd**, dan seterusnya.

Fitur biner ini memungkinkan Anda untuk mengekspresikan informasi dengan cara numerik, tanpa menyiratkan urutan apa pun.


Periksa kolom terakhir yang akan Anda kodekan.

Data dalam kolom **aspiration** hanya memiliki dua nilai: *std* dan *turbo*. Anda bisa mengkodekan kolom ini menjadi dua fitur biner. Namun, Anda juga dapat mengabaikan nilai *std* dan merekam apakah itu *turbo* atau bukan. Untuk melakukannya, Anda masih akan menggunakan metode `get_dummies` ini, tetapi tentukan `drop_first` sebagai *Betul*.


```python
df_car['aspiration'].value_counts()
```

```python
df_car = pd.get_dummies(df_car,columns=['aspiration'], drop_first=True)
```

```python
df_car.head()
```
